```python
### Importing libraries & data
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline 
sns.set()
```

```python
# قراءة الملف
import pandas as pd
df = pd.read_csv(r"C:\Users\hp\OneDrive\DISKTOP\Data Science Diploma\Data\day.csv")

# تحويل التاريخ
df['dteday'] = pd.to_datetime(df['dteday'])
df['day_name'] = df['dteday'].dt.day_name()
```

```python
df
```

```python
# Daily rental demand across working days vs. weekend
# نمط الطلب اليومي خلال أيام العمل والعطلات
```

```python
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='day_name', y='cnt', hue='workingday', order=days_order, ci=None)
plt.title('Average Daily Rentals by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Average Rentals')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


```

```python
# Map numeric codes to readable labels for better visualization
# تحويل الأكواد الرقمية إلى أسماء واضحة للرسم البياني.
```

```python
season_map = {1:'Spring',2:'Summer',3:'Fall',4:'Winter'}

df['season_name'] = df['season'].map(season_map)
```

```python
workingday_map = {1:'Work' , 0:'Holiday'}

df['workingday_name'] = df['workingday'].map(workingday_map)
```

```python
# Correlation matrix to identify relationships between weather features and rentals
# مصفوفة الارتباط لتحديد العلاقات بين عوامل الطقس ونسب التأجير
```

```python
plt.figure(figsize=(10, 8))

numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

```python
# Monthly total rentals comparing 2018 vs 2019 growth
# إجمالي التأجير الشهري لمقارنة النمو بين عامي 2018 و 2019
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(
    data=df, x='mnth', y='cnt', hue='yr', estimator='sum', ci=None, marker='o'
)
plt.title('Monthly Total Rentals: 2018 vs 2019')
plt.xlabel('Month')
plt.ylabel('Total Rentals')
plt.show()
```

```python
# Scatter plot showing rentals vs temperature grouped by weather severity
# تأثير درجة الحرارة وحالة الطقس على إجمالي التأجير
```

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='temp', y='cnt', hue='weathersit', palette='Set1')
plt.title('Rentals vs Temperature by Weather Situation')
plt.xlabel('Normalized Temperature')
plt.ylabel('Total Rentals')
plt.show()
```

```python
# Compare average daily rentals between casual and registered users across weekdays
# مقارنة متوسط التأجير اليومي بين المشتركين والعابرين عبر أيام الأسبوع
```

```python
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

user_types = df.groupby('day_name')[['casual', 'registered']].mean().reindex(days_order)

user_types.plot(kind='bar', figsize=(10, 5))
plt.title('Average Rentals by User Type Across Weekdays')
plt.xlabel('Day of Week')
plt.ylabel('Average Rentals')
plt.xticks(rotation=45)
plt.show()
```

```python
hottest_row = df[df['temp'] == df['temp'].max()]

windiest_row = df[df['windspeed'] == df['windspeed'].max()]

display(hottest_row[['dteday', 'temp', 'registered']])
display(windiest_row[['dteday', 'windspeed', 'registered']])
```
